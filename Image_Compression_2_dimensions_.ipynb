```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf

from tensorflow.keras import layers, Model


IMG_SIZE = 28
LATENT_DIM = 2
N_IMAGES = 6000

np.random.seed(42)
tf.random.set_seed(42)

print("TensorFlow Version:", tf.__version__)
print("Image Size:", IMG_SIZE, "x", IMG_SIZE)
print("Latent Dimensions:", LATENT_DIM)




def create_image():
    img = np.zeros((IMG_SIZE, IMG_SIZE), dtype=np.float32)

    # Randomly select a simple shape
    shape = np.random.randint(0, 3)

    # Random center
    cx = np.random.randint(8, 20)
    cy = np.random.randint(8, 20)

    if shape == 0:
        # Circle
        radius = np.random.randint(4, 8)

        y, x = np.ogrid[:IMG_SIZE, :IMG_SIZE]
        mask = (x - cx)**2 + (y - cy)**2 <= radius**2
        img[mask] = 1.0

    elif shape == 1:
        # Square
        size = np.random.randint(6, 12)

        x1 = max(0, cx - size // 2)
        x2 = min(IMG_SIZE, cx + size // 2)

        y1 = max(0, cy - size // 2)
        y2 = min(IMG_SIZE, cy + size // 2)

        img[y1:y2, x1:x2] = 1.0

    else:
        # Diagonal line
        thickness = np.random.randint(1, 3)

        for i in range(IMG_SIZE):
            j = i + np.random.randint(-3, 4)

            if 0 <= j < IMG_SIZE:
                img[max(0, j-thickness):min(IMG_SIZE, j+thickness+1), i] = 1.0

    # Add a small amount of noise
    noise = np.random.normal(0, 0.03, img.shape)

    img = img + noise
    img = np.clip(img, 0, 1)

    return img


# Generate images
X = np.array([create_image() for _ in range(N_IMAGES)])

# Add channel dimension
X = X.reshape(-1, IMG_SIZE, IMG_SIZE, 1)

# Train/Test split
split = int(0.8 * len(X))

X_train = X[:split]
X_test = X[split:]

print("\nDataset created successfully!")
print("Training images:", X_train.shape)
print("Testing images :", X_test.shape)


# ------------------------------------------------------------
# 3. DISPLAY ORIGINAL IMAGES
# ------------------------------------------------------------

plt.figure(figsize=(10, 4))

for i in range(10):
    plt.subplot(2, 5, i + 1)
    plt.imshow(X_train[i].squeeze(), cmap="gray")
    plt.axis("off")

plt.suptitle("Original 2D Images")
plt.show()


# ------------------------------------------------------------
# 4. BUILD ENCODER
# ------------------------------------------------------------

encoder_input = layers.Input(shape=(IMG_SIZE, IMG_SIZE, 1))

x = layers.Flatten()(encoder_input)

x = layers.Dense(256, activation="relu")(x)
x = layers.Dense(64, activation="relu")(x)

# 2-dimensional latent representation
latent = layers.Dense(LATENT_DIM, name="latent_space")(x)

encoder = Model(
    encoder_input,
    latent,
    name="Encoder"
)

encoder.summary()


# ------------------------------------------------------------
# 5. BUILD DECODER
# ------------------------------------------------------------

decoder_input = layers.Input(shape=(LATENT_DIM,))

x = layers.Dense(64, activation="relu")(decoder_input)
x = layers.Dense(256, activation="relu")(x)

x = layers.Dense(IMG_SIZE * IMG_SIZE, activation="sigmoid")(x)

decoder_output = layers.Reshape(
    (IMG_SIZE, IMG_SIZE, 1)
)(x)

decoder = Model(
    decoder_input,
    decoder_output,
    name="Decoder"
)

decoder.summary()


# ------------------------------------------------------------
# 6. COMBINE ENCODER + DECODER
# ------------------------------------------------------------

autoencoder_input = layers.Input(
    shape=(IMG_SIZE, IMG_SIZE, 1)
)

encoded = encoder(autoencoder_input)

decoded = decoder(encoded)

autoencoder = Model(
    autoencoder_input,
    decoded,
    name="Autoencoder"
)


# ------------------------------------------------------------
# 7. COMPILE MODEL
# ------------------------------------------------------------

autoencoder.compile(
    optimizer="adam",
    loss="mse"
)

print("\nAutoencoder Summary:")
autoencoder.summary()


# ------------------------------------------------------------
# 8. TRAIN THE AUTOENCODER
# ------------------------------------------------------------

history = autoencoder.fit(
    X_train,
    X_train,
    epochs=20,
    batch_size=128,
    validation_data=(X_test, X_test),
    verbose=1
)


# ------------------------------------------------------------
# 9. PLOT TRAINING LOSS
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

plt.plot(
    history.history["loss"],
    label="Training Loss"
)

plt.plot(
    history.history["val_loss"],
    label="Validation Loss"
)

plt.xlabel("Epoch")
plt.ylabel("MSE Loss")
plt.title("Autoencoder Training Loss")
plt.legend()
plt.grid()
plt.show()


# ------------------------------------------------------------
# 10. ENCODE TEST IMAGES
# ------------------------------------------------------------

encoded_images = encoder.predict(
    X_test,
    verbose=0
)

print("\nOriginal image dimensions:")
print("28 × 28 = 784 values")

print("\nCompressed representation:")
print("2 values =", encoded_images.shape[1], "latent dimensions")


# ------------------------------------------------------------
# 11. RECONSTRUCT IMAGES
# ------------------------------------------------------------

reconstructed_images = autoencoder.predict(
    X_test,
    verbose=0
)


# ------------------------------------------------------------
# 12. COMPARE ORIGINAL VS RECONSTRUCTED
# ------------------------------------------------------------

plt.figure(figsize=(12, 5))

for i in range(10):

    # Original
    plt.subplot(2, 10, i + 1)
    plt.imshow(
        X_test[i].squeeze(),
        cmap="gray"
    )
    plt.axis("off")

    # Reconstructed
    plt.subplot(2, 10, 10 + i + 1)
    plt.imshow(
        reconstructed_images[i].squeeze(),
        cmap="gray"
    )
    plt.axis("off")

plt.suptitle(
    "Top: Original Images | Bottom: Reconstructed Images"
)

plt.show()


# ------------------------------------------------------------
# 13. CALCULATE RECONSTRUCTION ERROR
# ------------------------------------------------------------

mse = np.mean(
    np.square(X_test - reconstructed_images)
)

print("\n--------------------------------")
print("RECONSTRUCTION RESULTS")
print("--------------------------------")

print("Mean Squared Error (MSE):", mse)


# ------------------------------------------------------------
# 14. CALCULATE PSNR
# ------------------------------------------------------------

psnr = tf.image.psnr(
    X_test,
    reconstructed_images,
    max_val=1.0
)

print(
    "Average PSNR:",
    float(tf.reduce_mean(psnr)),
    "dB"
)


# ------------------------------------------------------------
# 15. VISUALIZE 2D LATENT SPACE
# ------------------------------------------------------------

plt.figure(figsize=(8, 6))

plt.scatter(
    encoded_images[:, 0],
    encoded_images[:, 1],
    s=8,
    alpha=0.5
)

plt.xlabel("Latent Feature 1 (z1)")
plt.ylabel("Latent Feature 2 (z2)")

plt.title(
    "2D Latent Space of Autoencoder"
)

plt.grid()
plt.show()


# ------------------------------------------------------------
# 16. SHOW ONE IMAGE -> LATENT -> RECONSTRUCTION
# ------------------------------------------------------------

index = 0

original = X_test[index:index+1]

latent_value = encoder.predict(
    original,
    verbose=0
)

reconstructed = decoder.predict(
    latent_value,
    verbose=0
)

plt.figure(figsize=(12, 4))

# Original
plt.subplot(1, 3, 1)
plt.imshow(
    original[0].squeeze(),
    cmap="gray"
)
plt.title("Original 28×28")
plt.axis("off")

# Latent
plt.subplot(1, 3, 2)
plt.text(
    0.5,
    0.5,
    f"z1 = {latent_value[0][0]:.3f}\n"
    f"z2 = {latent_value[0][1]:.3f}",
    ha="center",
    va="center",
    fontsize=16
)
plt.title("2D Compressed Representation")
plt.axis("off")

# Reconstruction
plt.subplot(1, 3, 3)
plt.imshow(
    reconstructed[0].squeeze(),
    cmap="gray"
)
plt.title("Reconstructed 28×28")
plt.axis("off")

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 17. FINAL EXPERIMENT RESULT
# ------------------------------------------------------------

print("\n========================================")
print("FINAL EXPERIMENT RESULT")
print("========================================")

print("Input image       : 28 × 28 grayscale")
print("Original values   : 784")
print("Latent dimensions : 2")
print("Compression path  : 784 → 2 → 784")
print("MSE               :", mse)
print("Average PSNR      :", float(tf.reduce_mean(psnr)), "dB")

print("\nAutoencoder successfully learned")
print("a 2-dimensional representation")
print("and reconstructed the input images.")
```
