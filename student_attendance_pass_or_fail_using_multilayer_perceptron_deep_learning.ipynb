```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from sklearn.preprocessing import StandardScaler

X = np.array([
    [2, 60],
    [3, 70],
    [4, 90],
    [6, 80]
], dtype=float)

y = np.array([0, 0, 1, 1], dtype=float)

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

model = Sequential([
    Dense(8, activation='relu', input_shape=(2,)),
    Dense(4, activation='relu'),
    Dense(1, activation='sigmoid')
])

model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

model.fit(X_scaled, y, epochs=200, verbose=1)

loss, accuracy = model.evaluate(X_scaled, y, verbose=0)
print(f"\nAccuracy: {accuracy:.2f}")

new_students = np.array([
    [5, 80],
    [2, 85],
    [6, 75]
])

new_students_scaled = scaler.transform(new_students)

predictions = model.predict(new_students_scaled)

print("\nPredictions:")
for i, p in enumerate(predictions):
    label = 1 if p >= 0.5 else 0
    print(f"Student {i+1}: Probability={p[0]:.3f}, Predicted Result={label}")
```
