```python

x = [1, 2, 3, 4]
y = [5, 10, 15, 20]

w = 1.0
learning_rate = 0.1
batch_size = 2

print("Initial weight:", w)
print("=" * 50)


print("\nBATCH 1")

batch_x = x[0:2]
batch_y = y[0:2]

gradients = []

for xi, yi in zip(batch_x, batch_y):


    y_pred = w * xi


    gradient = (y_pred - yi) * xi

    gradients.append(gradient)

    print(f"x = {xi}, y = {yi}")
    print(f"Prediction = {y_pred}")
    print(f"Gradient = ({y_pred} - {yi}) * {xi} = {gradient}")
    print()


avg_gradient = sum(gradients) / batch_size

print("Average Gradient =", avg_gradient)


w = w - learning_rate * avg_gradient

print("Updated Weight =", w)


print("\n" + "=" * 50)
print("BATCH 2")

batch_x = x[2:4]
batch_y = y[2:4]

gradients = []

for xi, yi in zip(batch_x, batch_y):

    y_pred = w * xi

    gradient = (y_pred - yi) * xi

    gradients.append(gradient)

    print(f"x = {xi}, y = {yi}")
    print(f"Prediction = {y_pred}")
    print(f"Gradient = ({y_pred} - {yi}) * {xi} = {gradient}")
    print()

avg_gradient = sum(gradients) / batch_size

print("Average Gradient =", avg_gradient)

w = w - learning_rate * avg_gradient

print("Final Weight =", w)

print("\n" + "=" * 50)
print("FINAL RESULT")
print("Final weight =", w)
```

```python

w = 1.0

epochs = 1000

for epoch in range(epochs):


    batch_x = x[0:2]
    batch_y = y[0:2]

    gradients = []

    for xi, yi in zip(batch_x, batch_y):

        y_pred = w * xi

        gradient = (y_pred - yi) * xi

        gradients.append(gradient)

    avg_gradient = sum(gradients) / batch_size

    w = w - learning_rate * avg_gradient



    batch_x = x[2:4]
    batch_y = y[2:4]

    gradients = []

    for xi, yi in zip(batch_x, batch_y):

        y_pred = w * xi

        gradient = (y_pred - yi) * xi

        gradients.append(gradient)

    avg_gradient = sum(gradients) / batch_size

    w = w - learning_rate * avg_gradient



    if (epoch + 1) % 100 == 0:
        print(f"Epoch {epoch + 1}: w = {w}")




print("\n" + "=" * 50)
print("TRAINING COMPLETED")
print("=" * 50)

print("Total Epochs:", epochs)
print("Final Weight:", w)
```
